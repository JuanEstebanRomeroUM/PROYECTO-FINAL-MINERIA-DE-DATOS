### **2. Comprensión del conjunto de datos**
El conjunto de datos Online Retail II contiene un total reportado de 1.067.371 registros, distribuidos en dos periodos:

- Año 2010-2011: 541.910 registros.

Cada registro representa una línea de una factura, es decir, una operación asociada a un producto específico dentro de una transacción comercial. Por lo tanto, una misma factura puede contener varios registros cuando el cliente adquiere diferentes productos.

El conjunto de datos cuenta con ocho variables principales:

| InvoiceDate | Fecha y hora en que se realizó la transacción. |
| Price       | Precio unitario del producto expresado en libras esterlinas. |

Estas variables permiten estudiar diferentes dimensiones del negocio, como el comportamiento de compra de los clientes, el rendimiento de los productos, la evolución temporal de las ventas y la distribución geográfica de las transacciones.

Es importante considerar que los registros pueden contener valores faltantes, cantidades negativas, precios iguales a cero y operaciones canceladas, situaciones que deberán evaluarse durante las siguientes fases del proyecto.

## **FASE 4. Proceso ETL**

**Extract - Extraccion**

```python
# Copia de respaldo: no se modifica durante el ETL
df_original = df_final.copy(deep=True)

# Copia de trabajo para el proceso ETL
df_etl = df_original.copy(deep=True)

# Verificar estructura
print("Dimensiones de la copia original:", df_original.shape)
print("Dimensiones de la copia de trabajo:", df_etl.shape)

print("\nColumnas:")
print(df_etl.columns.tolist())

print("\nPrimeros registros:")
display(df_etl.head())
```

**Transform – Transformacion**

```python
# Estandarizar nombres de columnas
df_etl.columns = [
    "factura",
    "codigo_producto",
    "descripcion",
    "cantidad",
    "fecha_factura",
    "precio",
    "id_cliente",
    "pais",
    "periodo"
]

# Verificar nombres y tipos de datos
print("Columnas estandarizadas:")
print(df_etl.columns.tolist())

print("\nTipos de datos:")
print(df_etl.dtypes)

print("\nDimensiones:")
print(df_etl.shape)
```

```python
# Recuperar descripciones usando el código del producto
mapa_productos = (
    df_etl.dropna(subset=["descripcion"])
    .groupby("codigo_producto")["descripcion"]
    .agg(lambda x: x.mode().iloc[0])
)

df_etl["descripcion"] = df_etl["descripcion"].fillna(
    df_etl["codigo_producto"].map(mapa_productos)
)

# Verificar valores faltantes después del tratamiento
print("Valores faltantes después del tratamiento:")
print(df_etl.isnull().sum())

print("\nTotal de descripciones recuperadas:")
print(4382 - df_etl["descripcion"].isnull().sum())
```

La idea es recuperar descripciones únicamente cuando tenemos una referencia válida del mismo producto. Los clientes sin identificación permanecen en el conjunto.

```python
# Contar duplicados antes del tratamiento
duplicados_antes = df_etl.duplicated().sum()

# Eliminar duplicados exactos
df_etl = df_etl.drop_duplicates().copy()

# Verificar resultado
duplicados_despues = df_etl.duplicated().sum()

print("Duplicados antes:", duplicados_antes)
print("Duplicados eliminados:", duplicados_antes - duplicados_despues)
print("Duplicados después:", duplicados_despues)
print("Dimensiones actuales:", df_etl.shape)
```

Regla aplicada: eliminar registros completamente duplicados para evitar que una misma transacción quede contabilizada más de una vez.

Tratamiento de fehcas y cancelaciones 

```python
# Asegurar formato de fecha
df_etl["fecha_factura"] = pd.to_datetime(
    df_etl["fecha_factura"],
    errors="coerce"
)

# Identificar facturas canceladas
df_etl["es_cancelacion"] = (
    df_etl["factura"].astype(str).str.startswith("C")
)

# Verificar fechas y cancelaciones
print("Fechas inválidas:", df_etl["fecha_factura"].isna().sum())

print("\nRegistros de cancelación:")
print(df_etl["es_cancelacion"].value_counts())

print("\nCantidad de facturas canceladas diferentes:")
print(
    df_etl.loc[df_etl["es_cancelacion"], "factura"].nunique()
)

print("\nRango de fechas:")
print("Fecha inicial:", df_etl["fecha_factura"].min())
print("Fecha final:", df_etl["fecha_factura"].max())
```

## **FASE 6. Formulacion del problema de Machine Learning**

¿Qué deseamos predecir y por qué es útil?

Nuestro objetivo será desarrollar un modelo de clasificación binaria que identifique facturas asociadas a cancelaciones dentro del conjunto de datos Online Retail II.

Esto puede ser útil para que una empresa de comercio electrónico identifique y analice las operaciones canceladas, mida su frecuencia y encuentre patrones relacionados con productos, países, fechas y características de las transacciones.

Variable objetivo (y)
La variable objetivo será es_cancelacion.

Se construirá utilizando el prefijo de la columna original Invoice:
- 1: factura cuyo identificador comienza por C.
- 0: factura cuyo identificador no comienza por C.

Ejemplo:
| Invoice | es_cancelacion |
|---|---:|
| 536365 | 0 |
| C536379 | 1 |
| 536370 | 0 |
| C536383 | 1 |

Variables predictoras (X)

Para evitar que el modelo conozca directamente la respuesta, excluiremos Invoice de las variables predictoras.

| Variable | Descripción |
|---|---|
| `codigo_producto` | Código del producto |
| `descripcion` | Descripción del producto |
| `cantidad` | Cantidad registrada |
| `precio` | Precio unitario |
| `pais` | País de la operación |
| `fecha_factura` | Fecha y hora del registro |
| `periodo` | Periodo de origen de los datos |

Unidad de análisis

La unidad de análisis será la factura, no cada fila individual del dataset.

Esto es importante porque el conjunto de datos contiene varias filas para una misma factura, ya que cada una representa un producto o línea de compra.

Por lo tanto, agruparemos los registros por factura para construir una fila por cada factura.

Selección de registros y separación de datos

Se utilizarán los registros del conjunto ETL que tengan un identificador de factura válido.

Se conservarán tanto las facturas canceladas como las no canceladas, porque ambas clases son necesarias para entrenar el clasificador.

Para separar los datos:
- Se utilizará una separación cronológica, tomando las facturas más antiguas para entrenamiento y las más recientes para prueba.
- Se evitará que una misma factura aparezca simultáneamente en entrenamiento y prueba.
- Se evaluará el desempeño mediante métricas como precisión, recall, F1-score y matriz de confusión.

Revisar los datos disponibles

Antes de construir las variables, vamos a revisar cuántas facturas tenemos y cuántas están canceladas después de la limpieza ETL.

```python
# Crear una etiqueta de cancelación
df_etl["es_cancelacion"] = (
    df_etl["factura"].astype(str).str.startswith("C")
).astype(int)

# Contar facturas únicas por clase
resumen_facturas = (
    df_etl.groupby("es_cancelacion")["factura"]
    .nunique()
    .reset_index(name="numero_facturas")
)

resumen_facturas["tipo"] = resumen_facturas["es_cancelacion"].map({
    0: "Facturas normales",
    1: "Facturas canceladas"
})

print(resumen_facturas)
```

Algunos identificadores de cancelación comienzan por C seguido de un número. Queremos saber si, al quitar esa letra, encontramos una factura normal con el mismo número.


```python
# Obtener identificadores de facturas canceladas sin la C
ids_canceladas = (
    df_etl.loc[df_etl["es_cancelacion"] == 1, "factura"]
    .astype(str)
    .str[1:]
    .unique()
)

# Obtener identificadores de facturas normales
ids_normales = (
    df_etl.loc[df_etl["es_cancelacion"] == 0, "factura"]
    .astype(str)
    .unique()
)

# Buscar coincidencias
coincidencias = set(ids_canceladas).intersection(set(ids_normales))

print("Facturas canceladas:", len(ids_canceladas))
print("Facturas normales:", len(ids_normales))
print("Coincidencias entre ambas:", len(coincidencias))
print("Ejemplos de coincidencias:", list(coincidencias)[:10])
```

- No podemos utilizar Invoice como predictor porque revela directamente la etiqueta.
- No debemos utilizar cantidad negativa para afirmar que anticipamos una cancelación.
- Tampoco podemos asegurar que las demás variables representen información disponible antes del evento.

Separamos entrenamiento y prueba

```python
# Ordenar las facturas cronológicamente
df_facturas_ordenadas = df_facturas.sort_values(
    "fecha_factura"
).reset_index(drop=True)

# Calcular punto de separación
punto_corte = int(len(df_facturas_ordenadas) * 0.80)

# Dividir entrenamiento y prueba
df_entrenamiento = df_facturas_ordenadas.iloc[:punto_corte].copy()
df_prueba = df_facturas_ordenadas.iloc[punto_corte:].copy()

print("Facturas para entrenamiento:", len(df_entrenamiento))
print("Facturas para prueba:", len(df_prueba))

print("\nPeriodo de entrenamiento:")
print(df_entrenamiento["fecha_factura"].min(),
      "hasta",
      df_entrenamiento["fecha_factura"].max())

print("\nPeriodo de prueba:")
print(df_prueba["fecha_factura"].min(),
      "hasta",
      df_prueba["fecha_factura"].max())
```

Revisar la distribución de las clases

```python
print("DISTRIBUCIÓN EN ENTRENAMIENTO")
print(df_entrenamiento["es_cancelacion"].value_counts())
print(
    df_entrenamiento["es_cancelacion"]
    .value_counts(normalize=True).mul(100).round(2)
)

print("\nDISTRIBUCIÓN EN PRUEBA")
print(df_prueba["es_cancelacion"].value_counts())
print(
    df_prueba["es_cancelacion"]
    .value_counts(normalize=True).mul(100).round(2)
)
```
